## 5.Find the season with the highest average temperature range

```python
import pandas as pd
url="https://raw.githubusercontent.com/vega/vega-datasets/main/data/seattle-weather.csv"
weather_df = pd.read_csv(url)
weather_df['date'] = pd.to_datetime(weather_df['date'])

def get_season(month):
    if month in [12, 1, 2]:
        return 'Winter'
    elif month in [3, 4, 5]:
        return 'Spring'
    elif month in [6, 7, 8]:
        return 'Summer'
    else:
        return 'Autumn'

weather_df['season'] = weather_df['date'].dt.month.apply(get_season)

weather_df['temp_range'] = weather_df['temp_max'] - weather_df['temp_min']
    
avg_range_by_season = weather_df.groupby('season') ['temp_range'].mean() 

season_highest = avg_range_by_season.idxmax()
highest_value = avg_range_by_season.max()

print("Average temperature range by season:")
print(avg_range_by_season)
print()
print("Season with the highest average temperature range:", season_highest)
print("Highest average temperature range:", round(highest_value, 2))
```

## 6.Enhance input to accept months names

```python
import pandas as pd
url="https://raw.githubusercontent.com/vega/vega-datasets/main/data/seattle-weather.csv"
weather_df = pd.read_csv(url)
weather_df['date'] = pd.to_datetime(weather_df['date'])

weather_df['month_name'] = weather_df['date'].dt.month_name()
def get_season_from_name(month):
    month = month.lower()
    if month in ['December', 'January', 'February']:
        return 'Winter'
    elif month in ['March', 'April', 'May']:
        return 'Spring'
    elif month in ['June', 'July', 'August']:
        return 'Summer'
    else:
        return 'Autumn'

weather_df['season'] = weather_df['month_name'].apply(get_season_from_name)

weather_df['temp_range'] = weather_df['temp_max'] - weather_df['temp_min']
avg_range_by_season = weather_df.groupby('season') ['temp_range'].mean()

season_highest = avg_range_by_season.idxmax()
highest_value = avg_range_by_season.max()

print("Average temperature range by season:")
print(avg_range_by_season)
print()
print("Season with the highest average temperature range:", season_highest)
print("Highest average temperature range:", round(highest_value, 2))
```

## 7.Create a pie chart and histogram for the last column

```python
import matplotlib.pyplot as plt
weather_counts = weather_df['weather'].value_counts()
plt.figure(figsize=(6,6))
plt.pie(weather_counts, labels=weather_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Distribution of weather Types')
plt.show()

plt.figure(figsize = (7,5))
weather_df['weather'].value_counts().plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Frequency of Weather Types')
plt.xlabel('Weather Type')
plt.ylabel('Count')
plt.show()
```
