# Matplotlib

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

## Line Graph

```python
# Specifying the points
x = [1,2,3,4]
y = [2,4,6,8]

plt.plot(x,y, label = 'y=2x', marker = '*', markersize = 8, markeredgecolor = 'red')
plt.title('My First Graph')
plt.ylabel('Y Axis')
plt.xlabel('X Axis')
plt.xticks(x)

plt.legend()
plt.show()
```

## Bar Graph

```python
a = ['A', 'B', 'C']
b = [1, 4, 2]

bar1 = plt.bar(a,b)
plt.title('My Second Graph')
plt.ylabel('Y Axis')
plt.xlabel('X Axis')
plt.yticks(y)

bar1[0].set_hatch('/')
bar1[1].set_hatch('o')
bar1[2].set_hatch('*')
plt.figure(figsize = (3,8))
plt.show()
```
